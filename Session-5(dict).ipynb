## ***Dictionary Method***

#### *get()*

```python
my_dict.get("age")

# Docstring: Return the value for key if key is in the dictionary, else default.
```

#### *keys()*

```python
my_dict.keys()

# Docstring: Return a set-like object providing a view on the dict's keys. return all keys my dict
```

#### *values()*

```python
my_dict.values()

# Docstring: Return an object providing a view on the dict's values.
```

#### *items()*

```python
my_dict.items()

## return all key & values both in tuple form 
```

#### *pop()*

```python
my_dict.pop("age")

# docs =  remove specified key and return the corresponding value.
```

```python
my_dict
# remove krdia age ko
```

#### *update()*

```python
my_dict.update("color":"green")
# dict ha kurli bracket nh diye is err
```

```python
my_dict.update({"color":"green"})

# white sy green ma updtae kia
```

```python
my_dict
```

#### **Mannualy item change/update krny ka tariqa** 

```python

my_dict["color"] = "Green"
```

```python
my_dict  # again check kra
```

#### *fromkeys()*

```python
my_dict.fromkeys("Education", False)

# Signature: my_dict.fromkeys(iterable, value=None, /)
# Docstring: Create a new dictionary with keys from iterable and values set to value.
# ye loop ha ek trhn sy: e ko ek value samjh lia or false print kia aesy he all variable tk print krega
```

```python
my_dict
```

#### *popitem()*

```python
my_dict.popitem()
# Docstring:
# Remove and return a (key, value) pair as a 2-tuple.
```

```python
my_dict
```

#### *setdefault()*

```python
my_dict.setdefault('hobby', 'cricket')

# Insert key with a value of default if key is not in the dictionary
# agr dict ma key nh hoe tw default set krega end ma
```

```python
my_dict
```

```python
my_newDict
```

## ***Nested Dictionary***

```python
company = {
    "HR": {"manager": "Fatima", "employees": {"E1": "Ali","E2": "Usman"}},
    "IT": {"manager": "Khalid", "employees": {"E3": "Sara","E4": "Osama"}},
    "Finance": {"manager": "Zara", "employees": {"E5": "Hassan","E6": "Nadia"}}
}
```

```python
company
```

```python
# nested dict ma nested ma nested dict ha

company["IT"]["employees"]
```

```python
company["IT"]["employees"]["E3"]
```

```python
tech_company = {
    # LEVEL 1: Departments
    "Software Development": {
        "head": "Asif",
        
        # LEVEL 2: Teams inside Department
        "teams": {
            "AI Team": {
                "lead": "Bilal",
                
                # LEVEL 3: Projects handled by the Team
                "projects": {
                    "Chatbot AI": {
                        "budget": 50000,
                        
                        # LEVEL 4: Active Members and their Tech Stack inside Project
                        "members": {
                            "E101": {"name": "Zainab", "role": "ML Engineer", "skills": ["Python", "PyTorch"]},
                            "E102": {"name": "Hamza", "role": "Data Scientist", "skills": ["SQL", "Pandas"]}
                        }
                    },
                    "Vision App": {
                        "budget": 35000,
                        "members": {
                            "E103": {"name": "Ayesha", "role": "CV Specialist", "skills": ["OpenCV", "C++"]}
                        }
                    }
                }
            },
            "Web Team": {
                "lead": "Sana",
                "projects": {
                    "E-Commerce Portal": {
                        "budget": 80000,
                        "members": {
                            "E201": {"name": "Omer", "role": "Backend Developer", "skills": ["Django", "PostgreSQL"]}
                        }
                    }
                }
            }
        }
    }
}
tech_company
```

```python
# Har level ki key likhte jayenge
role = tech_company["Software Development"]["teams"]["AI Team"]["projects"]["Chatbot AI"]["members"]["E101"]["role"]
print(role)  # Output: ML Engineer
```

```python
# Akhir mein list ka index [0] lagaya
first_skill = tech_company["Software Development"]["teams"]["AI Team"]["projects"]["Chatbot AI"]["members"]["E101"]["skills"][0]
print(first_skill)  # Output: Python
```
