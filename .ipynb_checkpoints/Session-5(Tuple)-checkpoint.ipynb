```python

# Python Data Types Comparison Chart

# Feature        | Tuple                | List                 | Set                     | Dictionary
# --------------------------------------------------------------------------------------------------------
# Order          | Ordered              | Ordered              | Unordered               | Ordered (Python 3.7+)
# Mutable        | No (Immutable)       |Yes (Mutable)/chanable| Yes (Mutable)           | Yes (Mutable)
# Syntax         | (1, 2, 3)            | [1, 2, 3]            | {1, 2, 3}               | {'a':1, 'b':2}
# Duplicates     | Allowed              | Allowed              | Not Allowed             | Keys: Not Allowed
# Indexing       | Yes                  | Yes                  | No                      | Keys se access
# Key-Value Pair | No                   | No                   | No                      | Yes
# Speed          | Fast                 | Medium               | Fast                    | Fast
# Use Case       | Fixed data           | Changeable data      | Unique values           | Key-Value data
```
