```python
#    = = = = = = = = = = = = = = =   difference ()    = = = = = = = = = == = = = =


vip = {"Ali", "Sara", "Hamza", "Zainab", "Omer"}
recent_buyers = {"Hamza", "Zainab", "Bilal", "Nadia", "Ali"}

print(vip)
print(recent_buyers)
```

```python
vip.difference(recent_buyers)

# Return a new set with elements in the set that are not in the others
# First set mein jo values hain lekin second set mein nahi, woh deta hai.
# bilal is lye nh dia vip ka diff tha 
```

```python
vip
```

```python
#    = = = = = = = = = = = = = = =   difference_update ()    = = = = = = = = = == = = = =

vip.difference_update(recent_buyers)

# Update the set, removing elements found in others.
```

```python
recent_buyers
```

```python
#    = = = = = = = = = = = = = = =   isdisjoint ()    = = = = = = = = = == = = = =

mobile = {'Samsung', 'Vivo', 'Oppo'}

laptop = {'Dell', 'HP', 'Lenovo'}

print(mobile.isdisjoint(laptop))
```

```python
mobile = {'Samsung', 'Vivo', 'Sony'}         # EXAM 2

TV = {'panasonic', 'Sony', 'Lenovo'}

TV.isdisjoint(mobile)
	# Check karta hai ke dono sets mein koi common value nahi hai. retu Boolean ans
```

```python
#    = = = = = = = = = = = = = = =   issubset ()    = = = = = = = = = == = = = =

class_A = {'Anas', 'Ali', 'Ahmed'}

school = {'Anas', 'Ali', 'Ahmed', 'Usman', 'Hamza'}

print(class_A.issubset(school))

# Docstring: Report whether another set contains this set.
# 	Check karta hai ke ek set ki saari values doosre set mein mojood hain ya nahi.
```

```python
#                                     EXAMP 2 

school = {'Anas', 'Ali', 'Usman', 'Hamza'}
class_A = {'Anas', 'Ali', 'Ahmed'}

class_A.issubset(school) 
# class the ahmed school ma nh ha is lye false 
```

```python
#    = = = = = = = = = = = = = = =   issuperset ()    = = = = = = = = = == = = = =

school = {'Anas', 'Ali', 'Ahmed', 'Usman', 'Hamza'}

class_A = {'Anas', 'Ali', 'Ahmed'}

print(school.issuperset(class_A))

# Check karta hai ke doosre set ki saari values pehle set mein mojood hain ya nahi.
# Docstring: Report whether this set contains another set.
```

```python

Nested_Sets = {
    'Imran Khan', 'Bhutto',

    frozenset(cricketers),

    frozenset(my_No),

    408, 144, 109,

    frozenset(kitchen_home_appliance),

    frozenset(fruitSet),

    'mango', 'papaya', 'kiwi', 'graps', 'banana'
}

print(Nested_Sets)
```

```python
for players in Nested_Sets:
    if isinstance(players, frozenset):
        if 'Babar' in players:
            print('Babar')
            
```
